```python
import numpy as np

# Calculate velocity with rho_1,rho_2,A_1,A_2 and c_1 given (CONTINUITY)
def velocity_cnti(rho_1,rho_2,A_1,A_2,c_1):
    return (rho_1*A_1*c_1)/(rho_2*A_2)

# Calculate velocity with p_1,p_2,rho_1,rho_2 and c_1 given (BERNOULLI)
def velocity_blli(p_1,p_2,rho_1,rho_2,c_1):
    return np.sqrt( 2 * ((p_1/rho_1) - (p_2/rho_2)) + c_1**2)

# Calculate pressure (BERNOULLI)
def pressure_blli(p_1,rho_1,rho_2,c_1,c_2):
    return (p_1 + (rho_1*c_1**2)/2 - (rho_2*c_2**2)/2)

# Calculate pressure loss for a three-surface conical showerhead injector (INCOMPRESSIBLE)
# c_0 and p_0 must be given ( at inlet of injector )
def dpinj_incompr(A_0, A_1, A_2, c_0, p_0, rho):

    # Calculate velocity for (1) & (2)
    c_1 = A_0/A_1 * c_inj0
    c_2 = A_2/A_1 * c_1
    
    # Calculate pressures
    p_1 = pressure_blli(p_0, rho, rho, c_0, c_1)
    p_2 = pressure_blli(p_1, rho, rho, c_1, c_2)

    dp = p_0 - p_2
    
    return dp

def main():
    # 0: inj inlet, 1: inj orifice inlet, 2: inj outlet
    p_inj0 = 65 #bar, injector inlet
    p_inj2 = 55 #bar, injector outlet
    m_inj2  = 1 # kg/s
    rho = 700 # kg/m³, for incompressible case
    A_3 = 0.00025 # m²

    c_inj2 = m_inj2 / ( rho * A_3 )
    
    c_inj0 = velocity_blli(p_inj0, p_inj2, rho, rho, c_inj2) # m/s
    print(c_inj0)

main()
```
